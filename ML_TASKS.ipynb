```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a DataFrame for 10 students
data = {
    'Roll Number': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Ali', 'Ayesha', 'Ahmed', 'Fatima', 'Hassan',
             'Zainab', 'Usman', 'Maryam', 'Bilal', 'Hira'],
    'Department': ['BSAI', 'BSAI', 'BSCS', 'BSAI', 'BSCS',
                   'BSAI', 'BSCS', 'BSAI', 'BSCS', 'BSAI'],
    'Semester': [4, 4, 3, 4, 2, 4, 3, 4, 2, 4],
    'CGPA': [3.5, 3.8, 3.2, 3.9, 3.1, 3.7, 3.4, 3.6, 3.0, 3.85]
}

df = pd.DataFrame(data)

# Display the first five records
print("\nFirst Five Records:")
print(df.head())

# Display summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Create a bar chart of CGPA
plt.figure(figsize=(10, 5))
plt.bar(df['Name'], df['CGPA'])

plt.title('CGPA of 10 Students')
plt.xlabel('Student Name')
plt.ylabel('CGPA')
plt.ylim(0, 4.0)
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()

```
